```sql
-- Cell 1: 100% CORRECT & COMPLIANT SQL (Uppercase keywords, full descriptive names)
SELECT customer_id, customer_name, email
FROM workspace.customers
WHERE status = 'ACTIVE' AND created_date >= '2026-01-01'
ORDER BY customer_id ASC;
```

```sql
-- Cell 2: INCORRECT RULE VIOLATION (RULE-001: Lowercase & mixed-case keywords)
select o.order_id, o.amount, c.customer_name
fRoM workspace.orders o
left join workspace.customers c oN o.customer_id = c.customer_id
wHeRe o.status = 'COMPLETED' and o.amount > 100.0
order by o.amount desc;
```

```sql
-- Cell 3: REQUIRES LLM INTERVENTION (RULE-003 & RULE-004: SELECT * & Non-SARGable WHERE)
SELECT *
FROM workspace.large_clickstream_events
WHERE date_format(event_timestamp, 'yyyy-MM-dd') = '2026-08-10'
ORDER BY event_id;
```

```sql
-- Cell 4: REQUIRES LLM INTERVENTION (RULE-007: Cryptic Short-Forms/Abbreviations: cust, amt, txn, qty)
SELECT cust_id AS cust, SUM(txn_amt) AS amt, COUNT(txn_id) AS txn_cnt, AVG(item_qty) AS avg_qty
FROM workspace.daily_txns
WHERE txn_dt >= '2026-01-01'
GROUP BY cust_id;
```
