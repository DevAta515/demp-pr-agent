```sql
SELECT user_id, user_display_name, contact_email
FROM analytics_vault.registered_users
WHERE account_status = 'VERIFIED' AND signup_timestamp >= '2026-03-15'
ORDER BY user_id DESC;
```

```sql
select txn.transaction_id, txn.gross_amount, usr.display_name
fRoM sales_warehouse.transactions txn
left join sales_warehouse.users usr oN txn.user_id = usr.user_id
wHeRe txn.payment_status = 'PAID' and txn.gross_amount > 250.0
order by txn.gross_amount desc;
```

```sql
SELEC *
FROM staging_lakehouse.raw_web_activity
WHERE DATE_FORMAT(click_timestamp, 'yyyy-MM-dd') = '2026-09-01'
ORDER By click_id;
```

```sql
SELEC usr_id AS usr, SUM(txn_amt) AS total_amt, COUNT(txn_id) AS txn_cnt, AVG(prod_qty) AS avg_qty
FROM transaction_store.daily_ledger
WHERE ledger_dt >= '2026-04-01'
GROUP BY usr_id;
```
